```python
import random
import datetime

# Define constants for device types, locations, and reading ranges
DEVICE_TYPES = ['Temperature', 'Humidity', 'Pressure']
LOCATIONS = ['Building A, Room 101', 'Building B, Lab 203', 'Building C, Office 305']
READING_RANGES = {
    'Temperature': (20.0, 30.0),  # Celsius
    'Humidity': (40.0, 60.0),    # Percentage
    'Pressure': (950.0, 1050.0)  # hPa
}
```

```python
raw_sensor_data = []

for i in range(100):
    # timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
    timestamp = datetime.datetime.now() - datetime.timedelta(minutes=i)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    # device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
    device_id = f'DEV{i+1:03d}'

    # device_type: Randomly chosen from DEVICE_TYPES.
    device_type = random.choice(DEVICE_TYPES)

    # location: Randomly chosen from LOCATIONS.
    location = random.choice(LOCATIONS)

    # value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    raw_sensor_data.append((timestamp_str, device_id, device_type, location, value))
```

```python
# Display the first 5 simulated sensor readings
print("First 5 simulated sensor readings:")
for reading in raw_sensor_data[:5]:
    print(reading)
```

```python
temperature_readings = []

for reading in raw_sensor_data:
    timestamp_str, device_id, device_type, location, value = reading
    if device_type == 'Temperature':
        temperature_readings.append({
            'timestamp': timestamp_str,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        })

# Display the first 5 temperature readings to verify
print("First 5 temperature readings:")
for temp_reading in temperature_readings[:5]:
    print(temp_reading)
```

```python
unique_locations = set()

for reading in raw_sensor_data:
    # The location is the fourth element in the tuple (index 3)
    location = reading[3]
    unique_locations.add(location)

# Display the unique locations
print("Unique locations from raw_sensor_data:")
print(unique_locations)
```

```python
device_stats = {}

# Initialize device_stats with sensible large/small values
for device_type in DEVICE_TYPES:
    device_stats[device_type] = {'min_value': float('inf'), 'max_value': float('-inf')}

for reading in raw_sensor_data:
    _, _, device_type, _, value = reading

    # Update min_value
    if value < device_stats[device_type]['min_value']:
        device_stats[device_type]['min_value'] = value

    # Update max_value
    if value > device_stats[device_type]['max_value']:
        device_stats[device_type]['max_value'] = value

# Convert to the requested tuple format
final_device_stats = {dtype: (stats['min_value'], stats['max_value']) for dtype, stats in device_stats.items()}

# Display the device_stats
print("Device statistics (min_value, max_value) for each device type:")
print(final_device_stats)
```

```python
def get_readings_by_location(readings_list, target_location):
    """
    Filters a list of sensor readings to return only those from a specified location.

    Args:
        readings_list (list): A list of sensor reading tuples.
        target_location (str): The location to filter by.

    Returns:
        list: A list of sensor reading tuples for the target_location.
    """
    filtered_readings = []
    for reading in readings_list:
        # The location is the fourth element in the tuple (index 3)
        if reading[3] == target_location:
            filtered_readings.append(reading)
    return filtered_readings

# Demonstrate the function
target_location = 'Building A, Room 101'
readings_in_location_a = get_readings_by_location(raw_sensor_data, target_location)

print(f"\nReadings in '{target_location}': (first 5 shown if available)\n")
for reading in readings_in_location_a[:5]: # Displaying first 5 for brevity
    print(reading)
```

```python
def calculate_average_temperature(temp_data_list):
    """
    Calculates the average temperature from a list of temperature dictionaries.

    Args:
        temp_data_list (list): A list of dictionaries, each containing
                                a 'temperature_celsius' key.

    Returns:
        float: The average temperature, or 0.0 if the list is empty.
    """
    if not temp_data_list:
        return 0.0

    total_temperature = sum(item['temperature_celsius'] for item in temp_data_list)
    return total_temperature / len(temp_data_list)

# Demonstrate the function
average_temp = calculate_average_temperature(temperature_readings)
print(f"\nAverage temperature from all temperature readings: {average_temp:.2f} °C")

# Demonstrate with an empty list
empty_temp_list = []
average_temp_empty = calculate_average_temperature(empty_temp_list)
print(f"Average temperature from an empty list: {average_temp_empty:.2f} °C")
```

```python
def generate_report_line(reading_dict):
    """
    Formats a single temperature reading dictionary into a readable report line.

    Args:
        reading_dict (dict): A dictionary containing 'timestamp', 'device_id',
                             'location', and 'temperature_celsius' keys.

    Returns:
        str: A formatted string representing the temperature reading.
    """
    timestamp = reading_dict['timestamp']
    device_id = reading_dict['device_id']
    location = reading_dict['location']
    temperature = reading_dict['temperature_celsius']
    return f"[{timestamp}] Device {device_id} at {location}: {temperature}°C"

# Demonstrate the function using the first temperature reading
if temperature_readings:
    sample_reading = temperature_readings[0]
    report_line = generate_report_line(sample_reading)
    print(f"\nSample report line:\n{report_line}")
else:
    print("No temperature readings available to generate a report line.")
```
