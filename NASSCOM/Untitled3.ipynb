```python
import numpy as np
u=np.array([2,3])
v=np.array([4,1])
print("u+v=",u+v)
print("3*u=",3*u)
```

```python
import numpy as np
Maths=np.array([30,29,28,27])
Science=np.array([25,26,27,28])
English=np.array([27,26,28,25])
print("Maths: ",Maths)
print("Science: ",Science)
print("English: ",English)
print("Maths+English=",Maths+English)
print("Maths+Science=",Maths+Science)
print("English+Science=",English+Science)
print("Maths+Science+English=",Maths+Science+English)
```

```python
import numpy as np
A=np.array([[9,9],[9,9]])
B=np.array([[1,4],[2,5]])
print("A+B=\n",A+B)
print("A-B=\n",A-B)
print("A*B=\n",A*B)
print("A/B=\n",A/B)
print("A@B=\n",A@B)
```
